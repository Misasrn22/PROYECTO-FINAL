# CONCLUSIÓN

Ya con los resultados de mi red neuronal a mano, y las comparaciones que hemos hecho con scikit, podemos inferir varias cosas.

No podemos determinar al 100% la calidad de un vino por sus características fisicoquímicas. Las redes aciertan alrededor de 75%-77%, que es un 13% mas que si para todos los vinos diriamos "son buenos". Esto quiere decir que esos aspectos ayudan bastante, pero no son 100% determinantes, o por lo menos no con la cantidad de datos que tenemos.

De nuestros datos podemos decir:
- el corte en 6 deja las clases parejas (63% y 37%) aparte de ser la mediana esa tambien fue una de las razones por elegir ese numero.
- los atipicos, estuvo bien en no eliminarlos ya que todas eran mediciones reales y no errores.
- normalizar fue necesario. Se hicieron pruebas sin normalizar y daba 63% (lo mismo que decir siempre "bueno")

Con nuestros datos ya limpios vemos que las caracteristicas mas influyentes son:
- alcohol y acidez volatil. Mas alcohol se asocia a mejor calidad, y mientras menos acidez volatil tambien se asocia a un mejor vino.
- en los vinos blancos, encontramos tres asociaciones claves, a menor densidad, cloruros y dioxido de azufre se denoto mejor calidad.
- en los tintos, el alcohol es aun mas influyente y tambien los sulfatos, mientras mas se encontro mejor calidad en los paladares de los expertos.
- muchos de estos compuestos o caracteristicas quimicas estan asociados a la produccion. Un ejemplo es el dioxido de azufre, que actua como conservante. Podemos pensar una analogia, si quiero agregar menos dioxido de azufre, lo que debo hacer es cuidar mas la materia prima para que no deba agregar tanto para protegerla pues esta en buen estado. Los cloruros por su parte, provienen del suelo y el agua donde se riega, por lo que no puedo interferir quimicamente ahi. La acidez, viene de como fue la fermentacion.

Nuestra red nos arrojo que:
- llegamos a un valor parecido al scikit, y no hubo overfiting tampoco, lo que nos dice que fuimos por buen camino.
- la matriz de confusion nos mostro que la red detecta mejor los vinos buenos que los malos. Acierta 8 de cada 10 buenos y 6 de cada 10 malos. Esto puede ser debido a la limitacion de los datos, hay mas cantidad de buenos que de malos y por eso tiene mas datos asociados a ellos.

Las pruebas con scikit arrojaron que:
- lo optimo era usar 10 neuronas para nuestra red a mano.
- mas capas y distintas configuraciones no mejoraron el test, con lo cual alcanzamos un limite claro en 75%-77%

Pudimos ver como a pesar de pensar que tenemos bastante cantidad de datos (mas de seis mil) para poder clasificar algo mas objetivamente nos podria ayudar: mas cantidad de vinos y mas variables, aunque sin probarlo no podemos afirmarlo. Nuestra red nos dice que clasifica bien alrededor del 76% de los vinos, tambien podemos decir que se equivoca en 1 de cada 4 vinos. Nos puede servir como un pantallazo o una base para una primera impresion sin cata, pero luego la cata es necesaria para poder determinar con un poco mas de firmeza.

# Parte 3: Red en SCIKIT

En esta parte probaremos nuestra misma red en scikit. Con la facilidad que implica hacerla en scikit, tambien vamos a probar distintas alternativas, distintas neuronas, mas tasa de aprendizaje o menos. y comprobaremos que la opcion elegida (10 neuronas y tasa de 0.01) es la mejor.

***A) Primeros vamos con como era nuestra red original***

```python
#10 neuronas, tasa= 0.01
import pandas as pd
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
import matplotlib.pyplot as plt

df = pd.read_csv('vinos_limpio.csv')

# extraer variables de entrada (todas las filas, todas las columnas menos "calidad")
X = df.drop(columns=["Calidad"])

# Extraer columna de salida
Y = df["Calidad"]

# Separar los datos de entrenamiento y prueba
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=1/3, random_state=30, stratify=Y)

nn_10_01 = MLPClassifier(solver='sgd',
                      hidden_layer_sizes=(10, ),
                      activation='relu',
                      max_iter=100000,
                      learning_rate_init=.01,
                      random_state=30)

nn_10_01.fit(X_train, Y_train)

p_train = nn_10_01.predict_proba(X_train)[:, 1]
p_test = nn_10_01.predict_proba(X_test)[:, 1]

print("Puntaje del conjunto de entrenamiento: %f" % nn_10_01.score(X_train, Y_train))
print("Puntaje del conjunto de prueba: %f" % nn_10_01.score(X_test, Y_test))
print("epocas:", nn_10_01.n_iter_)
print("Error cuadrático medio train:", np.mean((p_train - np.array(Y_train))**2))
print("Error cuadrático medio test:", np.mean((p_test - np.array(Y_test))**2))
```

***B) Ya hecha nuestra red con scikit podemos ver algunas similitudes y diferencias.***

***Como similitudes claramente en la arquitectura de la red vemos descenso de gradiente estocastico, funcion de activacion relu, 1 capa oculta, 10 neuronas y tasa de aprendizaje de 0.01. El resultado que dio para los dos conjuntos (77,7% y 77,5$) es aproximado al que obtuvimos con nuestra red a mano (redondeando fue 75,5% ambos conjuntos). En las dos redes no padecemos de overfit, los dos conjuntos quedan casi iguales. Ambas redes llegan a un valor de precision y se estancan ahi, lo que nos da a pensar cosas para nuestra conclusion que luego veremos.***

***Como diferencias, la red de scikit rinde un poco mejor en los porcentajes de precision, y en el error cuadratico medio (0,152 y 0,155 contra 0,164 y 0,163 en la red a mano). Es una leve diferencia pero existente al fin y al cabo.***

***Otra diferencia es cuanto entreno cada una. Scikit hizo 404 epocas y la red a mano 23 epocas. Es mucha diferencia y tambien tal vez explique la leve mejora de scikit. Haciendo hincapie en esto tambien vemos el tiempo que tardo cada una en ejecutar en una computadora en igualdad de condiciones a pesar de la gran diferencia de epocas (3,5 segundos en promedio scikit contra 25 segundos en promedio la red a mano).***

***Esta mejoria de scikit se debe a varias diferencias internas que tienen: scikit usa minilotes para calcular el ajuste, mi red usa un solo vino para ajustar. Scikit tiene un recurso llamado "momentum" que lo que hace es recordar donde venia ajustandose, para que el nuevo ajuste, tambien tenga algo del pasado, y los "movimientos" son mas suave. Asi tambien scikit, como funcion de costo, usa la "entropia cruzada" internamente, mientras que nos usamos "error cuadratico". Para explicar brevemente esto ultimo, vamos a decir que la entropia cruzada asigna valores sin piso ni techo para la probabilidad de que la red este equivocada, castiga mucho a las probabilidades bajas y favorece a las muy altas. En cambio el error cuadratico nunca pasa del 1, por lo que tiene menos rango para favorecer y castigar. Aunque no son comparables estos dos, asi es como trabaja internamente scikit y posiblemente es una de las razones de la leve mejoria, aunque deberia analizarse mas profundamente.***


***C) Ahora en cada celda siguiente probaremos distintas alternativas. En el output veremos los puntajes y veremos que las 3 alternativas con una capa oculta  que muestros dan peor que la elegida.***

```python
#3 neuronas, tasa=0.05
nn3_05 = MLPClassifier(solver='sgd',
                   hidden_layer_sizes=(3, ),
                   activation='relu',
                   max_iter=100000,
                   learning_rate_init=.05,
                   random_state=30)

nn3_05.fit(X_train, Y_train)

print("Puntaje del conjunto de entrenamiento: %f" % nn3_05.score(X_train, Y_train))
print("Puntaje del conjunto de prueba: %f" % nn3_05.score(X_test, Y_test))
```

***Ahora veremos otras comparaciones, 2 y 3 capas ocultas, y el algoritmo "Adam".***
